```python
from ultralytics import YOLO

# เลือกโมเดล pose (เร็ว ๆ ใช้ nano)
model = YOLO("yolo11n-pose.pt")  # หรือ "yolov8n-pose.pt"

# track จากกล้องเว็บแคม (source=0)
model.track(
    source=0,          # 0 = webcam, หรือใส่ path วิดีโอ
    show=True,         # แสดงหน้าต่างภาพพร้อมกระดูกและ track ID
    conf=0.4,          # confidence threshold
    iou=0.5,           # IoU สำหรับ tracking
    persist=True,      # ให้รักษา track ID ข้ามเฟรม
    verbose=False
)
```
